Investigation 3


```python
df_clean = df.copy()
```

```python
df_clean.isnull().sum()
```

```python
num_cols = df_clean.select_dtypes(include='number').columns

for col in num_cols:
    df_clean[col].fillna(df_clean[col].median(), inplace=True)
```

```python
cat_cols = df_clean.select_dtypes(include='object').columns

for col in cat_cols:
    df_clean[col].fillna(df_clean[col].mode()[0], inplace=True)
```

```python
df_clean.drop_duplicates(inplace=True)
```

```python
df_clean.duplicated().sum()
```

```python
df_clean.dtypes
```

```python
print(df_clean.columns.tolist())
```

```python
df_clean["reservation_date"] = pd.to_datetime(
    df_clean["reservation_date"],
    format="%d-%m-%Y"
)
```

```python
num_cols = df_clean.select_dtypes(include='number').columns

for col in num_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean[col] = df_clean[col].clip(lower, upper)
```

```python
# Average Daily Rate cannot be negative
df_clean = df_clean[df_clean["avg_daily_rate"] >= 0]

# Advance booking days cannot be negative
df_clean = df_clean[df_clean["advance_booking_days"] >= 0]

# Stay duration cannot be negative
df_clean = df_clean[df_clean["total_stay_nights"] >= 0]

# Number of guests cannot be negative
df_clean = df_clean[df_clean["total_guests"] >= 0]
```
